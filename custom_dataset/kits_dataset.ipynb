```python
seg_dir = "/home/kathy/dataset/LITS/clean/segmentation-0.nii/"
source_dir = "/home/kathy/dataset/LITS/origin_clean/segmentation-0.nii/"
seg_files = [seg_dir + file for file in os.listdir(seg_dir)]
source_files = [source_dir + file for file in os.listdir(source_dir)]
```
